# Interactive version

Answer the prompts to choose the gain `g` and the self-couplings / sizes of each population. Run this notebook from the repository root (`jupyter notebook`), or add it to `PYTHONPATH`.

```python
import sys
sys.path.insert(0, '..')   # lets this notebook (in misc/) import rnn_library from the repo root

import numpy as np
import matplotlib.pyplot as plt

from rnn_library import make_coupling_matrix, simulate, population_acfs, timescales
from rnn_library import rnn_functions as rnn
from rnn_library.plotting import plot_acfs_with_traces
```

```python
print('Specify the value of the gain parameter, g.')
g = rnn.ask_user_float()

print('Specify the number of distinct self-couplings.')
K = rnn.ask_user_integer()

N_values, s_values = [], []
for idx in range(K):
    print(f'Population {idx}:\n s{idx} =')
    s_values.append(rnn.ask_user_float())
    print(f'Population {idx}:\n N{idx} =')
    N_values.append(rnn.ask_user_integer())
```

```python
t_end = 1000.0      # length of the recorded trajectory (ms)
dt = 0.1            # sampling interval (ms)
transient = 50.0    # discarded initial time (ms)
max_lag = 5000      # number of ACF lags (steps of dt)

J_Matrix = make_coupling_matrix(N_values, s_values, g)
t, x = simulate(J_Matrix, t_end, dt=dt, transient=transient)

lags = np.arange(max_lag) * dt
acfs = population_acfs(x, N_values, max_lag=max_lag)
tau = timescales(lags, acfs)
print(f'network timescales (ms): {np.round(tau, 2)}')
```

```python
plot_acfs_with_traces(t, x, lags, acfs, N_values, s_values, tau)
plt.show()
```
